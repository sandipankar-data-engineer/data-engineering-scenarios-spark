## Business Partner

```python
df_business_partner = spark.read.format('delta').table('bikesales.`sales-md`.business_partners')
df_addresses = spark.read.format('delta').table('bikesales.`sales-md`.addresses')
```

```python
df_business_partner_silver = df_business_partner.select('PARTNERID', 'COMPANYNAME','PARTNERROLE', 'ADDRESSID', 'CURRENCY').join(df_addresses.select('ADDRESSID', 'REGION', 'COUNTRY'), df_business_partner.ADDRESSID == df_addresses.ADDRESSID, 'inner').withColumnRenamed('COMPANYNAME', 'NAME').drop('ADDRESSID')
df_business_partner_silver.limit(5).display()
```

```python
df_business_partner_silver.write.format('delta').mode('overwrite').saveAsTable('bikesales.silver.business_partners')
```
