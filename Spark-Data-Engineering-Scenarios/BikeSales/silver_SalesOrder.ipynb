## SalesOrder

```python
%sql
CREATE OR REPLACE TABLE bikesales.silver.sales_orders AS
SELECT SALESORDERID, PARTNERID, SALESORG, FISCVARIANT, SUBSTRING(FISCALYEARPERIOD,0,4) AS FISCALYEAR, SUBSTRING(FISCALYEARPERIOD,5,8) AS FISCALPERIOD, CURRENCY, GROSSAMOUNT, NETAMOUNT FROM bikesales.`sales-td`.sales_orders;
```

```python
%sql
CREATE OR REPLACE TABLE bikesales.silver.sales_order_items AS
SELECT SALESORDERID,PRODUCTID, SALESORDERITEM, CURRENCY,QUANTITYUNIT, QUANTITY, GROSSAMOUNT, NETAMOUNT, TAXAMOUNT
FROM bikesales.`sales-td`.sales_order_items;
```
