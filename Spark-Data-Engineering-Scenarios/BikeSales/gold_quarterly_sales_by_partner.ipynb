```python
%sql
USE bikesales.silver;
CREATE OR REPLACE TABLE bikesales.gold.quarterly_sales_by_partner AS
SELECT 
    YEAR(f.FISCALYEAR) AS sales_year,
    f.FISCALPERIOD AS sales_quarter,
    CONCAT(YEAR(f.FISCALYEAR), '-Q', f.FISCALPERIOD) AS year_quarter,
    bp.PARTNERID,
    bp.REGION,
    COUNT(DISTINCT f.SALESORDERID) AS total_orders,
    SUM(f.NETAMOUNT) AS total_revenue,
    ROUND(SUM(f.NETAMOUNT) / NULLIF(COUNT(DISTINCT f.SALESORDERID), 0), 2) AS avg_order_value
FROM bikesales.silver.sales_orders as f
INNER JOIN bikesales.silver.business_partners as bp 
    ON f.PARTNERID = bp.PARTNERID
GROUP BY 1, 2, 3, 4, 5;
```
