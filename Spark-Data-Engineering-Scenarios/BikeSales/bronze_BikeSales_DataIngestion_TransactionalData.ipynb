# Transactional Data Ingestion

## Sales Data Ingestion

```python
%sql
-- Ingesting SalesOrders data from the volume folder and creating a table in the database 
USE bikesales.`sales-td`;
CREATE OR REPLACE TABLE sales_orders AS
SELECT *, _metadata.file_name AS _file_name, current_timestamp() AS _ingest_timestamp
FROM 
read_files(
    '/Volumes/bikesales/sales-td/dbv/SalesOrders.csv',
    inferSchema => true,
    header => true
);

SELECT * FROM sales_orders LIMIT 5;
```

```python
%sql
-- Ingesting SalesOrderItems data from the volume folder and creating a table in the database 
USE bikesales.`sales-td`;
CREATE OR REPLACE TABLE sales_order_items AS
SELECT *, _metadata.file_name AS _file_name, current_timestamp() AS _ingest_timestamp
FROM
read_files(
    '/Volumes/bikesales/sales-td/dbv/SalesOrderItems.csv',
    inferSchema=>true,
    header=>true
);

SELECT * FROM sales_order_items LIMIT 5;
```
