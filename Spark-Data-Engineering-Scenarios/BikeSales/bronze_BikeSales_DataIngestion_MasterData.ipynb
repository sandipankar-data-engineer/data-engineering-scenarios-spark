# Master Data Ingestion

## Business Partner Ingestion

```python
%sql
-- Ingesting Business Partner data from the volume folder and creating a table in the database 
USE bikesales.`sales-md`;
CREATE OR REPLACE TABLE business_partners AS
SELECT *, _metadata.file_name AS _file_name, current_timestamp() AS _ingest_timestamp
FROM 
read_files(
    '/Volumes/bikesales/sales-md/dbv/BusinessPartners.csv',
    header => true,
    inferSchema => true
);

SELECT * FROM business_partners LIMIT 5;
```
