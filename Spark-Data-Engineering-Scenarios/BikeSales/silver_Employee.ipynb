## Employees

```python
df_employees = spark.read.format('delta').table('bikesales.`sales-md`.employees')
df_addresses = spark.read.format('delta').table('bikesales.`sales-md`.addresses')
```

```python
from pyspark.sql.functions import concat, lit, col
df_employees_silver = df_employees.select('EMPLOYEEID', 'NAME_FIRST', 'NAME_LAST', 'PHONENUMBER', 'EMAILADDRESS', 'ADDRESSID').withColumn('NAME', concat(col('NAME_FIRST'), lit(' '), col('NAME_LAST'))).join(df_addresses.select('ADDRESSID', 'REGION', "COUNTRY"), df_employees.ADDRESSID == df_addresses.ADDRESSID, 'inner').drop('ADDRESSID', 'NAME_FIRST', 'NAME_LAST')
```

```python
df_employees_silver.write.format('delta').mode('overwrite').saveAsTable('bikesales.silver.employees')
```
