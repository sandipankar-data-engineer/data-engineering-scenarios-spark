# Lakeflow Jobs

## 1. Lakeflow Jobs _Overview_

* Lakeflow Jobs allows you to orchestrate every type of workload within Databricks, including notebooks, SQL queries, dashboards, pipelines, and much more.

### Lakeflow Jobs Key Features:

* Unified analytics for Data, Analytics and AI
* Key Benefits:
    * Simple authoring
    * Actionable insights
    * Proven reliability
* Native to Databricks integrates seamlessly with Databricks  capabilities (data ingestion/transformation/processing engine(photon)/governance(unity catalog)/storage(delta lake)/data warehousing/ML)

## 2. Lakeflow Jobs Core Components

### Job
- A job is the primary resource for:
    - Scheduling
    - Coordinating
    - Running operations such as data processing, ETL, analytics and ML workloads within the Databricks environment.
- Job consists of one or more than one tasks.

#### Jobs Orchestration
* Jobs consist of one or more Tasks
![image_1794672498325.png](./image_1794672498325.png "image_1794672498325.png")
* Control flows can be established between Tasks
![image_1794672516650.png](./image_1794672516650.png "image_1794672516650.png")
* Jobs support different Triggers
![image_1794672531596.png](./image_1794672531596.png "image_1794672531596.png")
* The compute layer supports all these different orchestration patterns.


### Tasks
- A task is a single unit of work within a job that executes a specific workload such as :
    - Notebook
    - Script
    - Query and more
#### Task Types:
* Notebook Task (Source, Path, Compute options, etc.)
![image_1794672311678.png](./image_1794672311678.png "image_1794672311678.png")
* SQL Task (Task name, SQL query, SQL warehouse, etc.)
