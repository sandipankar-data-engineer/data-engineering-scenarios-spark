#### Compute Options
##### Interactive Clusters
- Can be shared by multiple users.
- Best suited for ad-hoc analysis, data exploration, and development.
- Can be convenient for interactive notebook execution.
- Generally not the preferred choice for production job workloads because compute may remain running when not needed.

##### Job Clusters
- Compute provisioned for job execution.
- Automatically terminate when the job finishes.
- Suitable for production workloads.
- Can reduce costs by avoiding idle compute.
- Subject to cloud-provider and cluster startup times.
- Multiple tasks in a job can reuse the same job cluster when configured to do so.

##### Serverless
- Fully managed compute.
- Reduces infrastructure management.
- Supports automatic scaling.
- Offers faster startup and built-in performance optimizations.
- Helps simplify operations and improve workload efficiency.

##### SQL Warehouse
- Purpose-built for SQL queries, dashboards, and BI workloads.
- Supports high concurrency and autoscaling.
- Provides Intelligent Workload Management.
- Supports auto-start and auto-stop.
- Cluster sizing can be adjusted to control costs.
- Serverless SQL warehouses are fully managed.

![image_1794675970813.png](./image_1794675970813.png "image_1794675970813.png")

##### Compute Comparison
|Compute|Best suited for|Key characteristics|
|---|---|---|
|Interactive Cluster|Development and exploration|Shared, interactive|
|Job Cluster|Production job workloads|Job-scoped, terminates after execution|
|Serverless|Managed job execution|Fully managed, auto-scaling, simpler operations|
|SQL Warehouse|SQL and BI workloads|SQL-optimized, concurrency and autoscaling|

##### Serverless Performance Mode

* Optimized mode enables faster job startup and execution, making it ideal for time-sensitive workloads. This setting applies only to tasks with serverless compute in your job.

#### Task Orchestration

##### What is a DAG?

* A DAG is a conceptual representation of series of activities, including data processing flows.
    * Directed - unambiguous direction for each edge
    * Acyclic - contains no cycles
    * Graph - collection of vertices connected by edges


##### Task Orchestration Overview

* Databricks Jobs supports task orchestration
* Run multiple tasks - The ability to run multiple tasks as a Directed Acyclic Graph (DAG).
* Orchestrate tasks - Orchestrate tasks using the Databricks UI, API, SDK, or Declarative Automation Bundles.
* You can define the order of execution of tasks in a job by configuring task dependencies, creating a DAG of task execution.

##### Control Flow

###### Retries

- Specifies how many times a failed task should be re-executed.
- Useful for handling transient failures (e.g., network interruptions or temporary service errors).
- Automatically restarts the task up to the configured retry limit.
- Some Databricks features, such as Structured Streaming schema evolution, rely on retries to restart failed executions.
- Continuous Jobs automatically retry using exponential backoff.

![image_1794766284850.png](./image_1794766284850.png "image_1794766284850.png")

###### Run if Conditional Tasks

- Controls whether a downstream task executes based on the outcome of its upstream tasks.
- Supports the following conditions:
|Condition|Meaning|
|---|---|
|All succeeded|All upstream tasks succeeded|
|At least one succeeded|One or more upstream tasks succeeded|
|None failed|No upstream task failed|
|All done|All upstream tasks completed, regardless of outcome|
|At least one failed|One or more upstream tasks failed|
|All failed|All upstream tasks failed|

![image_1794766721330.png](./image_1794766721330.png "image_1794766721330.png")

###### If/else Conditional Tasks

- Enables branching logic based on a condition.
- Conditions can use:
  - Task Values (taskValues)
  - Job Parameters
  - Dynamic Value References
- Supports the following comparison operators:
==, !=, >, >=, <, <=

![image_1794766948596.png](./image_1794766948596.png "image_1794766948596.png")

###### For each Tasks

- Executes a nested task repeatedly using different input parameters.
- Consists of two components:
    - For each task: Defines the loop and input collection.
    - Nested task: Defines the operation to execute in each iteration.
- Supports passing different parameters to the nested task for each iteration.
- Useful for processing multiple tables, files, or datasets using the same logic.

![image_1794767713435.png](./image_1794767713435.png "image_1794767713435.png")

###### Disabled Tasks

- Allows a task to be skipped during Job execution without deleting it.
- Retains the task's configuration and execution history.
- Downstream tasks evaluate their configured Run if conditions to determine whether they should execute.

### 2. Parameters

Parameters are key-value pairs used to pass inputs to Jobs and Tasks, making workflows reusable and adaptable.

#### Task Parameters

- Key-value pairs defined at the individual Task level.
- Used to pass inputs to a specific task.
- Allow the same task logic to behave differently based on input values.
- Support advanced orchestration:
    - Conditional execution: Control workflow branches using input values.
    - Looping: Supply arrays or input collections to For each tasks.
    - Task configuration: Pass environment-specific or processing-specific inputs. <br>

**Setting Task Parameter:**
- Open the required Task.
- Navigate to its Parameters configuration.
- Add the required key-value pairs.
![image_1794814605002.png](./image_1794814605002.png "image_1794814605002.png")

#### Job Parameters

* Key-value pairs defined at the Job level.
* Provide common input values to the entire workflow.
* Automatically pushed down to compatible tasks.
* Task parameters with the same key are overridden by Job parameters.

**Setting Job Parameter:**
* Open the Job.
* Navigate to the Parameters section.
* Add key-value pairs.

#### Accessing Parameters in Tasks

* Parameters can be accessed using dbutils.widgets.get().
* Example:
<br>`environment = dbutils.widgets.get("environment")`
<br>`processing_date = dbutils.widgets.get("processing_date")`

#### Task Values – Passing Information Between Tasks

##### Defining Task Values

Use dbutils.jobs.taskValues.set() to create a key-value pair.
```

dbutils.jobs.taskValues.set(
    key="catalog_name",
    value="dbacademy"
)
```

##### Accessing Task Values

Use dbutils.jobs.taskValues.get() to retrieve a value from another task.
```
catalog = dbutils.jobs.taskValues.get(
    taskKey="task-name",
    key="catalog_name"
)
```

#### Dynamic Value References

- Dynamic Value References allow you to access runtime information from Job and Task contexts without hardcoding values. 
- Values are resolved dynamically during Job execution.
- Avoid hardcoding runtime information.
- Support passing task outputs to downstream tasks.
- Useful for dynamic paths, logging, execution tracking, and conditional workflows.
<br>

**Syntax** <br>
Use double curly braces {{ }} to reference dynamic values.

Types of Dynamic Value References
|Reference|Purpose|
|---|---|
|{{job.run_id}}|Unique Job run identifier|
|{{job.start_time.day}}|Day of Job execution|
|{{job.parameters.environment}}|Job-level parameter value|
|{{task.name}}|Current task name|
|{{task.retry_count}}|Number of retry attempts|
|{{tasks.data-validation.values.record_count}}|Retrieve record count|
|{{tasks.file-processor.values.output_path}}|Retrieve generated output path|


#### Job Parameters vs Task Parameters

|Aspect|Job Parameters|Task Parameters|
|---|---|---|
|Definition|Job level|Individual Task level|
|Scope|Workflow-wide|Task-specific|
|Reusability|Shared across tasks|Specific to a task|
|Override behavior|Override same-key task parameters|Can be overridden by Job parameters|
|Runtime override|Supported|Depends on task configuration and parameter mapping|

#### Notification Events
|Event|Description|
|---|---|
|On Start|Job or Task execution begins|
|On Success|Job or Task completes successfully|
|On Failure|Job or Task fails|
|On Duration Warning|Execution exceeds a configured duration threshold|
|On Streaming Backlog|Streaming workload experiences backlog|

#### Notification Destinations
- Email
- Slack
- Microsoft Teams
- PagerDuty
- Webhooks

#### Webhooks
- Allow integration with external applications and third-party services.
- Send execution-related notifications to configured endpoints.
- Useful for custom alerting systems and incident management workflows.

#### Designing Notification Strategies
- Configure Job-level alerts for overall workflow monitoring.
- Configure Task-level alerts for critical processing steps.
- Notify on failures to enable quick investigation.
- Use duration warnings to identify unusually long-running Jobs.
- Use streaming backlog notifications to identify processing delays.
- Use different notification destinations depending on operational requirements.

### 4. Job Schedules and Triggers

##### What is a Trigger?
* Triggers help automate workflows without requiring manual execution.

##### Types of Triggers
|Trigger|Description|Typical Use Case|
|---|---|---|
|Scheduled|Executes at predefined times or intervals|Daily ETL|
|File Arrival|Executes when new files arrive|Event-driven ingestion|
|Continuous|Keeps a Job running continuously|Streaming workloads|
|Manual|Executes when explicitly initiated|Debugging and backfills|
|Table Update|Executes when monitored tables are updated|Event-driven data processing|

##### Scheduled Trigger
* Automatically executes Jobs at predefined times or intervals.
* Supports recurring schedules such as hourly, daily, or weekly.
* Can be configured through the Databricks UI.
* Supports cron expressions for more flexible scheduling.
* Use cases:
    - Daily ETL pipelines.
    - Hourly data synchronization.
    - Periodic reporting and aggregations.
