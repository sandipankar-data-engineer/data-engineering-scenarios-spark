# Databricks Lakeflow

## What is Lakeflow in Databricks?

- Lakeflow is Databricks' unified data engineering solution for building, orchestrating, and managing data pipelines.
- It is a platform with **optimized storage** (_Delta Lake/Parquet/Iceberg_), **unified governance** with _Unity Catalog_ and powered by Lakeflow to deliver end-to-end, high quality data engineering for analytics and AI.
- Lakeflow is an end-to-end data engineering solution within Databricks.
- It helps manage the complete data engineering lifecycle:
    - Extract data from source systems.
    - Ingest data into the Lakehouse.
    - Transform raw data into structured datasets.
    - Apply data quality checks.
    - Orchestrate multiple pipelines and tasks
    - Monitor, troubleshoot, and manage production workflows.

## Lakeflow components

| Component | Purpose | Description |
|---|---|---|
| Lakeflow Connect | Data ingestion | Ingest from databases, SaaS applications, files and other sources |
| Lakeflow Spark Declarative Pipelines (SDP) | Data transformation | Build batch and streaming ETL pipelines |
| Lakeflow Jobs | Orchestration | Schedule and orchestrate notebooks, pipelines, SQL and other tasks |
| Lakeflow Designer | Visual development | Build data transformation workflows visually |
