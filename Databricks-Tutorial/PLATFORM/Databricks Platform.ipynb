# 1. High-level architecture


## 1. Databricks objects

* A Databricks account is the top-level construct that you use to manage Databricks across your organization. 
* At the account level, you manage:
    * Identity and access
    * Workspace management
    * Unity Catalog metastore management
    * Usage management
* An account can contain multiple workspaces and Unity Catalog metastores.

## 2. Workspaces

* Workspaces are the collaboration environment where users run compute workloads such as ingestion, interactive exploration, scheduled jobs, and ML training.

### Workspace architecture

**Control Plane**<br>
The control plane includes the backend services that Databricks manages in your Databricks account. The control plane is located in the Databricks account, not your cloud account. The web application is in the control plane.

**Compute Plane**<br>
The compute plane is where your data is processed. There are two types of compute planes depending on the compute that you are using.

For serverless compute, the serverless compute resources run in a serverless compute plane in your Databricks account.

For classic Databricks compute, the compute resources are in your Google Cloud resources in what is called the classic compute plane. This refers to the network in your Google Cloud resources and its resources.

### Classic workspace architecture

Classic Databricks workspaces have three associated storage buckets known as the workspace storage buckets. The workspace storage buckets are in your Google Cloud account.

![image_1795187481364.png](./image_1795187481364.png "image_1795187481364.png")

### Serverless workspace architecture
Workspace storage in serverless workspaces is stored in the workspace's default storage. You can also connect to your cloud storage account to access your data.

## Unity Catalog

* Unity Catalog metastores are the central governance system for data assets such as tables and ML models. You organize data in a metastore under a three-level namespace:
```<catalog-name>.<schema-name>.<object-name>```




# 2. Compute

## 1. Serverless compute

* **Fully managed compute**
* Databricks handles infrastructure provisioning, startup, scaling, optimization, and maintenance.
* Users can focus on workloads without manually configuring the underlying compute infrastructure.
* Particularly useful when teams want minimal infrastructure administration.
* Main Purpose: Simplify compute management and provide fast, automatically managed resources.
* Example:
    * Running notebooks
    * Automated jobs
    * Lakeflow pipelines 
    * SQL analytics
    * Supported AI/ML workloads

## 2. Classic Compute

* Compute where users or administrators have greater control over configuration, such as Databricks Runtime, instance/VM types, worker count, autoscaling settings, libraries, and policies.
* Main Purpose: Provide configurable compute for workloads requiring **greater infrastructure control**.
* Example:
    * Custom Spark workloads
    * Interactive development
    * Specialized libraries
    * Legacy workloads, or workloads requiring configurations unavailable with serverless compute

## 3. SQL Warehouses

* Specialized compute optimized for executing SQL queries and supporting analytics workloads. 
* SQL Warehouses are separate from general-purpose Spark notebook compute and are designed for SQL performance and concurrency.
* Main Purpose: Provide high-performance SQL analytics and BI query processing.
* Example:
    * Databricks SQL Editor
    * Dashboards
    * Reporting
    * Ad-hoc SQL analysis
    * BI tools such as Power BI and Tableau


## 4. Jobs Compute


* Compute used to execute automated Databricks jobs and workflows.
* Compute can be provisioned for a job run and terminated when the workload completes, rather than remaining continuously active.
* Main Purpose: Execute scheduled and production workloads efficiently.
* Example:
    * ETL/ELT jobs
    * Scheduled data transformations
    * Batch processing
    * Data-quality jobs
    * Workflow tasks
    * Production Spark applications.

## 5. All-Purpose Compute

* Interactive compute intended primarily for development and exploratory work. 
* It can be attached to notebooks so users can interactively execute Python, PySpark, SQL, Scala, or R workloads depending on the environment.
* Main Purpose: Provide an interactive environment for development, exploration, and debugging.
* Example:
    * Developing PySpark code
    * Exploring datasets
    * Testing transformations
    * Debugging applications
    * Running notebooks
    * Performing ad-hoc analysis.


## 6. Lakeflow Declarative Pipelines Compute

* Compute used by Lakeflow Declarative Pipelines to execute declaratively defined batch and streaming data pipelines. 
* Databricks manages pipeline execution based on the selected compute configuration.
* Main Purpose: Build and operate reliable production data pipelines.
* Example:
    * Bronze-Silver-Gold architectures
    * Streaming ingestion
    * Batch transformations
    * Incremental processing
    * Data-quality expectations
    * Managed ETL/ELT pipelines

## 7. Model Serving Compute


* Managed compute behind Databricks Model Serving endpoints. 
* It provides scalable infrastructure for serving ML models, foundation models, and supported GenAI/agent applications through real-time endpoints.
* Main Purpose: Make trained models and AI applications available for real-time inference.
* Example:
    * Real-time predictions
    * ML model APIs
    * GenAI applications
    * RAG applications
    * AI agents
    * Applications consuming Databricks serving endpoints.
